```python
num_cols = ['quantity', 'unit_price','discount', 'delivery_days', 'customer_rating', 'revenue']
for col in num_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(),bins=10)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['quantity'],
           df['revenue'])
plt.title('Quantity vs Revenue')
plt.xlabel('Quantity')
plt.ylabel('Revenue')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['discount'],
           df['revenue'])
plt.title('Discount vs Revenue')
plt.xlabel('Discount')
plt.ylabel('Revenue')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['customer_rating'],
           df['revenue'])
plt.title('Customer Rating vs Revenue')
plt.xlabel('Customer Rating')
plt.ylabel('Revenue')
plt.show()
```

```python
sns.pairplot(df[['quantity', 'unit_price', 'discount', 'revenue']])
plt.show()
```

```python
category_sales = df.groupby('product_category')['revenue'].sum()
print(category_sales)
```

```python
category_quantity = df.groupby('product_category')['quantity'].sum()
print(category_quantity)
```

```python
region_sales = df.groupby('region')['revenue'].sum()
print(region_sales)
```

```python
payment_analysis= df.groupby('payment_method')['revenue'].sum()
print(payment_analysis)
```

```python
pivot_table = pd.pivot_table(df,
                             values='revenue',
                            index='product_category',
                            columns='region',
                            aggfunc='sum')
print(pivot_table)
```

```python
num_cols = ['quantity', 'unit_price', 'discount', 'delivery_days', 'customer_rating', 'revenue']
correlation = df[num_cols].corr()
print(correlation)
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
print(df[num_cols].mean())
```

```python
print(df[num_cols].median())
```

```python
print(df[num_cols].std())
```

```python
category_sales.plot(kind='bar', figsize=(8,5))
plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Revenue')
plt.show()
```

```python
payment_counts = df['payment_method'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%')
plt.title('Payment Method Distribution')
plt.show()
```

```python
monthly_sales = df.groupby('order_month') ['revenue'].sum()
monthly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.show()
```
